```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

actual = test_data["UNITS"]
predicted = test_data["NAIVE_FORECAST"]

mae = mean_absolute_error(actual, predicted)
rmse = np.sqrt(mean_squared_error(actual, predicted))
mape = np.mean(np.abs((actual - predicted) / actual)) * 100

print("Naive Forecast Performance")
print("---------------------------")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("MAPE:", round(mape, 2), "%")
```

```python
# STEP 78 - Create Lag Features

ml_data = pd.concat([
    train_data[["WEEK_END_DATE", "UNITS"]],
    test_data[["WEEK_END_DATE", "UNITS"]]
]).reset_index(drop=True)

ml_data["LAG_1"] = ml_data["UNITS"].shift(1)
ml_data["LAG_2"] = ml_data["UNITS"].shift(2)
ml_data["LAG_4"] = ml_data["UNITS"].shift(4)

ml_data["ROLLING_4"] = (
    ml_data["UNITS"]
    .shift(1)
    .rolling(window=4)
    .mean()
)

ml_data = ml_data.dropna().reset_index(drop=True)

ml_data.head(10)
```

```python
# STEP 79 - Random Forest Forecasting Model

from sklearn.ensemble import RandomForestRegressor

features = [
    "LAG_1",
    "LAG_2",
    "LAG_4",
    "ROLLING_4"
]

# Split based on the original training/test boundary
train_ml = ml_data.iloc[:len(train_data) - 4].copy()
test_ml = ml_data.iloc[len(train_data) - 4:].copy()

X_train = train_ml[features]
y_train = train_ml["UNITS"]

X_test = test_ml[features]

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    max_depth=10
)

rf_model.fit(X_train, y_train)

rf_forecast = rf_model.predict(X_test)

test_ml["RF_FORECAST"] = rf_forecast

test_ml[[
    "WEEK_END_DATE",
    "UNITS",
    "RF_FORECAST"
]].head(10)
```

```python
# STEP 80 - Random Forest Performance

rf_mae = mean_absolute_error(
    test_ml["UNITS"],
    test_ml["RF_FORECAST"]
)

rf_rmse = np.sqrt(mean_squared_error(
    test_ml["UNITS"],
    test_ml["RF_FORECAST"]
))

rf_mape = np.mean(
    np.abs(
        (test_ml["UNITS"] - test_ml["RF_FORECAST"])
        / test_ml["UNITS"]
    )
) * 100

print("Random Forest Performance")
print("-------------------------")
print(f"MAE  : {rf_mae:.2f}")
print(f"RMSE : {rf_rmse:.2f}")
print(f"MAPE : {rf_mape:.2f}%")
```

```python
# STEP 81 - Random Forest Feature Importance

feature_importance = pd.DataFrame({
    "Feature": features,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

feature_importance
```

```python
# STEP 88: Generate future 13-week forecasts

import numpy as np
import pandas as pd

forecast_horizon = 13

# Start with all available actual values
history = list(ml_data["UNITS"])

# Last historical date
last_date = pd.to_datetime(ml_data["WEEK_END_DATE"].iloc[-1])

future_dates = []
future_forecasts = []

for i in range(forecast_horizon):
    
    # Create lag features
    lag_1 = history[-1]
    lag_2 = history[-2]
    lag_3 = history[-3]
    lag_4 = history[-4]
    lag_8 = history[-8]
    
    # Create rolling features
    rolling_4 = np.mean(history[-4:])
    rolling_8 = np.mean(history[-8:])
    
    # Prepare input for the model
    future_input = pd.DataFrame([{
        "LAG_1": lag_1,
        "LAG_2": lag_2,
        "LAG_3": lag_3,
        "LAG_4": lag_4,
        "LAG_8": lag_8,
        "ROLLING_4": rolling_4,
        "ROLLING_8": rolling_8
    }])
    
    # Predict next week
    prediction = final_rf_model.predict(future_input)[0]
    
    # Future date
    future_date = last_date + pd.Timedelta(days=7 * (i + 1))
    
    future_dates.append(future_date)
    future_forecasts.append(prediction)
    
    # Add prediction to history for recursive forecasting
    history.append(prediction)

# Create forecast table
future_forecast_df = pd.DataFrame({
    "WEEK_END_DATE": future_dates,
    "FORECASTED_UNITS": future_forecasts
})

future_forecast_df["FORECASTED_UNITS"] = future_forecast_df["FORECASTED_UNITS"].round(0).astype(int)

print("Future 13-week forecast generated successfully.")
print()
print(future_forecast_df)
```

```python
# Final 13-week product-store forecast with Random Forest
# Recursive forecasting using previous predictions

from sklearn.base import clone

# Start with the last available training observations
history = model_train[
    ["STORE_NUM", "UPC", "WEEK_END_DATE", "UNITS"]
].copy()

history = history.sort_values(
    ["STORE_NUM", "UPC", "WEEK_END_DATE"]
)

# Use test rows only for known future price/promotion/calendar information
future_info = model_test[
    [
        "STORE_NUM", "UPC", "WEEK_END_DATE",
        "PRICE", "BASE_PRICE",
        "FEATURE", "DISPLAY", "TPR_ONLY",
        "CATEGORY", "SUB_CATEGORY", "PRODUCT_SIZE",
        "SEG_VALUE_NAME",
        "PARKING_SPACE_QTY",
        "SALES_AREA_SIZE_NUM",
        "AVG_WEEKLY_BASKETS",
        "WEEK_OF_YEAR", "MONTH"
    ]
].copy()

forecast_results = []

# Work week-by-week
for week in sorted(test_weeks):

    week_data = future_info[
        future_info["WEEK_END_DATE"] == week
    ].copy()

    rows = []

    for _, row in week_data.iterrows():

        store = row["STORE_NUM"]
        upc = row["UPC"]

        # Historical + previously predicted demand
        pair_history = history[
            (history["STORE_NUM"] == store) &
            (history["UPC"] == upc)
        ].sort_values("WEEK_END_DATE")

        units_history = pair_history["UNITS"].values

        # Need previous 8 observations
        if len(units_history) < 8:
            continue

        # Build lag features
        lag_1 = units_history[-1]
        lag_2 = units_history[-2]
        lag_3 = units_history[-3]
        lag_4 = units_history[-4]
        lag_8 = units_history[-8]

        rolling_4 = np.mean(units_history[-4:])
        rolling_8 = np.mean(units_history[-8:])

        discount_pct = (
            (row["BASE_PRICE"] - row["PRICE"])
            / row["BASE_PRICE"]
            if row["BASE_PRICE"] > 0
            else 0
        )

        feature_row = pd.DataFrame([{
            "LAG_1": lag_1,
            "LAG_2": lag_2,
            "LAG_3": lag_3,
            "LAG_4": lag_4,
            "LAG_8": lag_8,
            "ROLLING_4": rolling_4,
            "ROLLING_8": rolling_8,
            "PRICE": row["PRICE"],
            "BASE_PRICE": row["BASE_PRICE"],
            "DISCOUNT_PCT": discount_pct,
            "FEATURE": row["FEATURE"],
            "DISPLAY": row["DISPLAY"],
            "TPR_ONLY": row["TPR_ONLY"],
            "UPC": row["UPC"],
            "CATEGORY": row["CATEGORY"],
            "SUB_CATEGORY": row["SUB_CATEGORY"],
            "PRODUCT_SIZE": row["PRODUCT_SIZE"],
            "STORE_NUM": row["STORE_NUM"],
            "SEG_VALUE_NAME": row["SEG_VALUE_NAME"],
            "PARKING_SPACE_QTY": row["PARKING_SPACE_QTY"],
            "SALES_AREA_SIZE_NUM": row["SALES_AREA_SIZE_NUM"],
            "AVG_WEEKLY_BASKETS": row["AVG_WEEKLY_BASKETS"],
            "WEEK_OF_YEAR": row["WEEK_OF_YEAR"],
            "MONTH": row["MONTH"]
        }])

        # Apply the same categorical encoding used during training
        for col in categorical_cols:
            combined_categories = pd.concat(
                [
                    model_train[col],
                    model_test[col]
                ]
            )

            # These columns are already encoded in model_train/model_test.
            # Convert future values using the training/test mapping.
            feature_row[col] = model_test.loc[
                (model_test["STORE_NUM"] == store) &
                (model_test["UPC"] == upc) &
                (model_test["WEEK_END_DATE"] == week),
                col
            ].iloc[0]

        # Prediction from all individual RF trees
        tree_predictions = np.array([
            tree.predict(feature_row[ml_features])[0]
            for tree in rf_product_store.estimators_
        ])

        prediction = tree_predictions.mean()
        lower = np.percentile(tree_predictions, 2.5)
        upper = np.percentile(tree_predictions, 97.5)

        # Avoid negative demand
        prediction = max(0, prediction)
        lower = max(0, lower)
        upper = max(0, upper)

        rows.append({
            "WEEK_END_DATE": week,
            "STORE_NUM": store,
            "UPC": upc,
            "FORECAST": prediction,
            "LOWER_95": lower,
            "UPPER_95": upper
        })

        # Add prediction to history for recursive forecasting
        history = pd.concat([
            history,
            pd.DataFrame([{
                "STORE_NUM": store,
                "UPC": upc,
                "WEEK_END_DATE": week,
                "UNITS": prediction
            }])
        ], ignore_index=True)

    forecast_results.extend(rows)

final_forecast = pd.DataFrame(forecast_results)

print("Final forecast rows:", len(final_forecast))
print("Forecast weeks:", final_forecast["WEEK_END_DATE"].nunique())

print("\nForecast period:")
print(
    final_forecast["WEEK_END_DATE"].min(),
    "to",
    final_forecast["WEEK_END_DATE"].max()
)

print("\nSample forecasts:")
print(final_forecast.head(10))
```

```python
# FAST FINAL 13-WEEK PRODUCT-STORE FORECAST
# Recursive forecasting + Random Forest uncertainty

from collections import deque

# --------------------------------------------------
# 1. Create history for each product-store pair
# --------------------------------------------------

history_dict = {}

history_source = (
    model_train[
        ["STORE_NUM", "UPC", "WEEK_END_DATE", "UNITS"]
    ]
    .sort_values(["STORE_NUM", "UPC", "WEEK_END_DATE"])
)

for (store, upc), group in history_source.groupby(
    ["STORE_NUM", "UPC"]
):
    history_dict[(store, upc)] = deque(
        group["UNITS"].tail(8).tolist(),
        maxlen=8
    )

# --------------------------------------------------
# 2. Future information
# --------------------------------------------------

future_info = model_test[
    [
        "STORE_NUM", "UPC", "WEEK_END_DATE",
        "PRICE", "BASE_PRICE",
        "FEATURE", "DISPLAY", "TPR_ONLY",
        "CATEGORY", "SUB_CATEGORY", "PRODUCT_SIZE",
        "SEG_VALUE_NAME",
        "PARKING_SPACE_QTY",
        "SALES_AREA_SIZE_NUM",
        "AVG_WEEKLY_BASKETS",
        "WEEK_OF_YEAR", "MONTH"
    ]
].copy()

forecast_results = []

# --------------------------------------------------
# 3. Recursive week-by-week forecasting
# --------------------------------------------------

for week in sorted(test_weeks):

    week_data = future_info[
        future_info["WEEK_END_DATE"] == week
    ].copy()

    # Build lag features from stored history
    lag_values = []

    for _, row in week_data.iterrows():

        key = (row["STORE_NUM"], row["UPC"])
        h = history_dict.get(key)

        if h is None or len(h) < 8:
            lag_values.append([np.nan] * 7)
            continue

        values = list(h)

        lag_values.append([
            values[-1],                 # LAG_1
            values[-2],                 # LAG_2
            values[-3],                 # LAG_3
            values[-4],                 # LAG_4
            values[-8],                 # LAG_8
            np.mean(values[-4:]),       # ROLLING_4
            np.mean(values[-8:])        # ROLLING_8
        ])

    lag_df = pd.DataFrame(
        lag_values,
        columns=[
            "LAG_1", "LAG_2", "LAG_3",
            "LAG_4", "LAG_8",
            "ROLLING_4", "ROLLING_8"
        ],
        index=week_data.index
    )

    # Combine features
    X_week = pd.concat(
        [
            lag_df,
            week_data[
                [
                    "PRICE",
                    "BASE_PRICE",
                    "FEATURE",
                    "DISPLAY",
                    "TPR_ONLY",
                    "UPC",
                    "CATEGORY",
                    "SUB_CATEGORY",
                    "PRODUCT_SIZE",
                    "STORE_NUM",
                    "SEG_VALUE_NAME",
                    "PARKING_SPACE_QTY",
                    "SALES_AREA_SIZE_NUM",
                    "AVG_WEEKLY_BASKETS",
                    "WEEK_OF_YEAR",
                    "MONTH"
                ]
            ]
        ],
        axis=1
    )

    # Discount percentage
    X_week["DISCOUNT_PCT"] = np.where(
        X_week["BASE_PRICE"] > 0,
        (X_week["BASE_PRICE"] - X_week["PRICE"])
        / X_week["BASE_PRICE"],
        0
    )

    # Put columns in exact training order
    X_week = X_week[ml_features]

    # Remove any impossible rows
    valid_mask = X_week.notna().all(axis=1)

    valid_X = X_week.loc[valid_mask]

    # --------------------------------------------------
    # Predictions from all trees
    # --------------------------------------------------

    tree_predictions = np.column_stack([
        tree.predict(valid_X)
        for tree in rf_product_store.estimators_
    ])

    predictions = tree_predictions.mean(axis=1)
    lower_95 = np.percentile(tree_predictions, 2.5, axis=1)
    upper_95 = np.percentile(tree_predictions, 97.5, axis=1)

    predictions = np.maximum(predictions, 0)
    lower_95 = np.maximum(lower_95, 0)
    upper_95 = np.maximum(upper_95, 0)

    # --------------------------------------------------
    # Save forecasts + update history
    # --------------------------------------------------

    valid_rows = week_data.loc[valid_mask].copy()

    for i, (_, row) in enumerate(valid_rows.iterrows()):

        store = row["STORE_NUM"]
        upc = row["UPC"]
        prediction = predictions[i]

        forecast_results.append({
            "WEEK_END_DATE": week,
            "STORE_NUM": store,
            "UPC": upc,
            "FORECAST": prediction,
            "LOWER_95": lower_95[i],
            "UPPER_95": upper_95[i]
        })

        # Add prediction for recursive next week
        history_dict[(store, upc)].append(prediction)

    print(
        week.strftime("%Y-%m-%d"),
        "→",
        len(valid_rows),
        "forecasts"
    )

# --------------------------------------------------
# 4. Final forecast dataframe
# --------------------------------------------------

final_forecast = pd.DataFrame(forecast_results)

print("\n================================")
print("FINAL FORECAST COMPLETE")
print("================================")

print("Forecast rows:", len(final_forecast))
print("Product-store pairs:",
      final_forecast[["STORE_NUM", "UPC"]].drop_duplicates().shape[0])
print("Forecast weeks:",
      final_forecast["WEEK_END_DATE"].nunique())

print("\nForecast period:")
print(
    final_forecast["WEEK_END_DATE"].min(),
    "to",
    final_forecast["WEEK_END_DATE"].max()
)

print("\nSample forecasts:")
print(final_forecast.head(10))
```

```python
# Complete missing product-store forecasts using historical-average fallback

# All required product-store-week combinations
required_forecasts = test_model[
    ["WEEK_END_DATE", "STORE_NUM", "UPC"]
].copy()

# Identify forecasts already produced by Random Forest
existing_keys = final_forecast[
    ["WEEK_END_DATE", "STORE_NUM", "UPC"]
].drop_duplicates()

# Find missing product-store-week combinations
missing_forecasts = required_forecasts.merge(
    existing_keys,
    on=["WEEK_END_DATE", "STORE_NUM", "UPC"],
    how="left",
    indicator=True
)

missing_forecasts = missing_forecasts[
    missing_forecasts["_merge"] == "left_only"
].drop(columns="_merge")

print("Missing forecast rows:", len(missing_forecasts))

# --------------------------------------------------
# Historical average and standard deviation
# --------------------------------------------------

historical_stats = (
    train_data
    .groupby(["STORE_NUM", "UPC"])["UNITS"]
    .agg(["mean", "std"])
    .reset_index()
)

missing_forecasts = missing_forecasts.merge(
    historical_stats,
    on=["STORE_NUM", "UPC"],
    how="left"
)

# Fallback forecast = historical average
missing_forecasts["FORECAST"] = (
    missing_forecasts["mean"].fillna(0)
)

# Approximate 95% uncertainty interval
missing_forecasts["LOWER_95"] = np.maximum(
    0,
    missing_forecasts["mean"] -
    1.96 * missing_forecasts["std"].fillna(0)
)

missing_forecasts["UPPER_95"] = (
    missing_forecasts["mean"] +
    1.96 * missing_forecasts["std"].fillna(0)
)

missing_forecasts = missing_forecasts[
    [
        "WEEK_END_DATE",
        "STORE_NUM",
        "UPC",
        "FORECAST",
        "LOWER_95",
        "UPPER_95"
    ]
]

# Add fallback forecasts to RF forecasts
final_forecast_complete = pd.concat(
    [
        final_forecast,
        missing_forecasts
    ],
    ignore_index=True
)

# Sort
final_forecast_complete = final_forecast_complete.sort_values(
    ["WEEK_END_DATE", "STORE_NUM", "UPC"]
).reset_index(drop=True)

# --------------------------------------------------
# Validation checks
# --------------------------------------------------

print("\n================================")
print("COMPLETE FINAL FORECAST")
print("================================")

print(
    "Forecast rows:",
    len(final_forecast_complete)
)

print(
    "Expected rows:",
    len(evaluation_pairs) * len(test_weeks)
)

print(
    "Product-store pairs:",
    final_forecast_complete[
        ["STORE_NUM", "UPC"]
    ].drop_duplicates().shape[0]
)

print(
    "Forecast weeks:",
    final_forecast_complete[
        "WEEK_END_DATE"
    ].nunique()
)

print("\nForecast period:")
print(
    final_forecast_complete["WEEK_END_DATE"].min(),
    "to",
    final_forecast_complete["WEEK_END_DATE"].max()
)

print("\nSample:")
print(final_forecast_complete.head(10))
```
